```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
#Upload the diamond dataset
import pandas as pd
data ='https://raw.githubusercontent.com/pmbrione/Skyline-Stats-C1000/refs/heads/main/Diamonds.csv'

diamond_df = pd.read_csv(data)
diamond_df
```

```python
sns.scatterplot(data=diamond_df, x='carat', y='price')
```

```python
sns.scatterplot(data=diamond_df, x='depth', y='price')
```

```python
Y = diamond_df['price']
X = diamond_df[['carat','depth']]
X = sm.add_constant(X)

reg_equation =sm.OLS(Y,X).fit()
print(reg_equation.summary())
```

```python
Y = diamond_df['price']
X = diamond_df[['carat','depth', 'x', 'y', 'z', 'table']]
X = sm.add_constant(X)

reg_equation =sm.OLS(Y,X).fit()
print(reg_equation.summary())
```
